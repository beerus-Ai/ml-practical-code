# Logistic Regression - German Credit Risk

## Import Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## Load Dataset

```python
df = pd.read_csv("german_credit_risk (1).csv")
df.head()
```

## Shape

```python
df.shape
```

## Drop Unnecessary Column

```python
df = df.drop("Unnamed: 0", axis=1)
df.head()
```

## Check Missing Values

```python
df.isnull().sum()
```

## Fill Missing Values

```python
cat_cols = ["Sex", "Housing", "Saving accounts",
            "Checking account", "Purpose"]

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

## Convert Categorical Variables

```python
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
df.head()
```

## Convert Target Variable

```python
df["Risk"] = df["Risk"].map({"bad": 0, "good": 1})
df["Risk"].value_counts()
```

## Features and Target

```python
X = df.drop("Risk", axis=1)
y = df["Risk"]
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y)

print(X_train.shape)
print(X_test.shape)
```

## Feature Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Train the Model

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)
```

## Prediction

```python
y_pred = model.predict(X_test_scaled)
y_pred[:10]
```

## Accuracy

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
```

## Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
```

## Classification Report

```python
print(classification_report(y_test, y_pred))
```

## New Customer

```python
new_customer = pd.DataFrame({
    "Age": [30],
    "Sex": ["male"],
    "Job": [2],
    "Housing": ["own"],
    "Saving accounts": ["little"],
    "Checking account": ["moderate"],
    "Credit amount": [4000],
    "Duration": [24],
    "Purpose": ["car"]
})

new_customer
```

## Preprocess New Customer

```python
new_customer = pd.get_dummies(new_customer, columns=cat_cols, drop_first=True)
new_customer = new_customer.reindex(columns=X.columns, fill_value=0)
new_customer.head()
```

## Predict Credit Risk

```python
new_customer_scaled = scaler.transform(new_customer)
prediction = model.predict(new_customer_scaled)

if prediction[0] == 1:
    print("Prediction: GOOD CREDIT RISK")
else:
    print("Prediction: BAD CREDIT RISK")
```

## Predict Probability

```python
probability = model.predict_proba(new_customer_scaled)
print(probability)
print("Probability of Good Risk:", probability[0][1])
```
