# Sentiment Analysis - TF-IDF + Logistic Regression

## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

## Load Dataset

```python
df = pd.read_csv("tweets_dataset.csv")
df.head()
```

## Shape

```python
df.shape
```

## Prepare the Data

```python
X = df["tweet"].fillna("")
y = df["sentiment"].str.lower().str.strip()

y.value_counts()
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y)

print("Training data:", len(X_train))
print("Testing data :", len(X_test))
```

## TF-IDF Vectorization

```python
tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Number of TF-IDF features:", X_train_tfidf.shape[1])
```

## Train the Model

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)

y_pred = model.predict(X_test_tfidf)
print("Model training completed.")
```

## Model Evaluation

```python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(y_test, y_pred, pos_label="positive", zero_division=0)
recall = recall_score(y_test, y_pred, pos_label="positive", zero_division=0)
f1 = f1_score(y_test, y_pred, pos_label="positive", zero_division=0)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

## Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Negative", "Positive"],
            yticklabels=["Negative", "Positive"])
plt.title("Sentiment Classification")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

## Word Cloud

```python
all_text = " ".join(df["tweet"].astype(str))

wordcloud = WordCloud(width=800, height=400,
                      background_color="white").generate(all_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Reviews")
plt.show()
```

## Positive and Negative Word Clouds

```python
positive_text = " ".join(df[df["sentiment"] == "positive"]["tweet"].astype(str))
negative_text = " ".join(df[df["sentiment"] == "negative"]["tweet"].astype(str))

positive_wordcloud = WordCloud(width=800, height=400,
                               background_color="white").generate(positive_text)
negative_wordcloud = WordCloud(width=800, height=400,
                               background_color="white").generate(negative_text)

plt.figure(figsize=(10, 5))
plt.imshow(positive_wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Positive Sentiment Word Cloud")
plt.show()

plt.figure(figsize=(10, 5))
plt.imshow(negative_wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Negative Sentiment Word Cloud")
plt.show()
```

## Predict a New Review

```python
new_review = ["I really enjoyed this product. It is excellent!"]

new_review_tfidf = tfidf.transform(new_review)
prediction = model.predict(new_review_tfidf)

print("New Review:", new_review[0])
print("Predicted Sentiment:", prediction[0])
```
